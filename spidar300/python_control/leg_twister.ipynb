```python
from servo import Servo
from time import sleep
```

```python
PORT = "/dev/ttyACM1"
s1 = Servo(port = PORT, servo_id=111, model="STS3215")
s1.ping()
s2 = Servo(port = PORT, servo_id=112, model="STS3215")
s2.ping()
s3 = Servo(port = PORT, servo_id=113, model="STS3215")
s3.ping()
```

```python
while True:
    pos, deg = s2.position()
    s2.move_to_degrees(deg + 20)
    pos, deg = s3.position()
    s3.move_to_degrees(deg - 40)

    sleep(0.5)

    pos, deg = s2.position()
    s2.move_to_degrees(deg - 20)
    sleep(0.5)
    pos, deg = s3.position()
    s3.move_to_degrees(deg + 40)

    sleep(0.5)
```
